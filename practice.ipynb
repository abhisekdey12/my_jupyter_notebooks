```python
def StingFunc(s):
    start = s
    new_s = ''
    for i in range(len(s)-1):
        if s[i].lower() == s[i+1].lower():
            new_s = s[:i] + s[i+2:]
            break
        else:
            new_s = s
    if new_s == s:
        return new_s
    else:        
        return StingFunc(new_s)
```

```python
r = StingFunc("abcdaadc")
if len(r) == 0:
    print("Empty String")
else:
    print(r)
```
